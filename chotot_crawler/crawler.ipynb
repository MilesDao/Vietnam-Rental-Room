```python
import requests
import json
import pandas as pd
import time
import random

# ==========================================
# CONFIG
# ==========================================

URL = "https://gateway.chotot.com/v1/public/ad-listing"

CG_LIST = [1050, 1010, 1020]

TARGET_DATA = 7000
LIMIT = 20

HANOI_REGION = 12000

OUTPUT_CSV = "chotot_hanoi_raw.csv"
OUTPUT_JSON = "chotot_hanoi_raw.json"


# ==========================================
# STORAGE
# ==========================================

all_ads = []
seen_ids = set()


# ==========================================
# CRAWL EACH CATEGORY
# ==========================================

for CG in CG_LIST:

    # Nếu đã đủ 3000 thì dừng
    if len(all_ads) >= TARGET_DATA:
        break

    print("\n" + "=" * 60)
    print(f"START CRAWLING CG = {CG}")
    print("=" * 60)

    offset = 0

    while len(all_ads) < TARGET_DATA:

        params = {
            "cg": CG,
            "limit": LIMIT,
            "o": offset,
            "st": "u",
            "region_v2": HANOI_REGION
        }

        try:
            response = requests.get(
                URL,
                params=params,
                timeout=20
            )

        except requests.RequestException as e:

            print("Request error:", e)
            print("Waiting 10 seconds...")

            time.sleep(10)
            continue


        print(
            f"CG={CG} | "
            f"offset={offset} | "
            f"status={response.status_code}"
        )


        # ======================================
        # 429
        # ======================================

        if response.status_code == 429:

            print("429 Too Many Requests!")
            print("Waiting 30 seconds...")

            time.sleep(30)
            continue


        # ======================================
        # OTHER ERROR
        # ======================================

        if response.status_code != 200:

            print("Request failed!")
            print(response.text[:300])

            print("Waiting 10 seconds...")
            time.sleep(10)

            continue


        # ======================================
        # PARSE RESPONSE
        # ======================================

        data = response.json()

        ads = data.get("ads", [])


        # ======================================
        # NO MORE ADS
        # ======================================

        if not ads:

            print(f"CG={CG}: No more ads.")
            break


        # ======================================
        # ADD UNIQUE ADS
        # ======================================

        new_ads = 0

        for ad in ads:

            ad_id = ad.get("ad_id")

            if ad_id is None:
                continue

            if ad_id not in seen_ids:

                seen_ids.add(ad_id)

                all_ads.append(ad)

                new_ads += 1


        print(
            f"Received: {len(ads)} | "
            f"New: {new_ads} | "
            f"Total: {len(all_ads)}"
        )


        # ======================================
        # TARGET REACHED
        # ======================================

        if len(all_ads) >= TARGET_DATA:

            print("\nTARGET REACHED!")
            break


        # ======================================
        # NEXT PAGE
        # ======================================

        offset += LIMIT


        # ======================================
        # DELAY
        # ======================================

        delay = random.uniform(1.5, 3.0)

        print(f"Waiting {delay:.1f}s...")

        time.sleep(delay)


# ==========================================
# SAVE RAW DATA
# ==========================================

print("\n" + "=" * 60)
print("CRAWLING FINISHED")
print("=" * 60)

print(f"Total unique ads: {len(all_ads)}")


# ==========================================
# SAVE JSON
# ==========================================

with open(
    OUTPUT_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        all_ads,
        f,
        ensure_ascii=False,
        indent=2
    )


# ==========================================
# SAVE CSV
# ==========================================

df_raw = pd.DataFrame(all_ads)

df_raw.to_csv(
    OUTPUT_CSV,
    index=False,
    encoding="utf-8-sig"
)


print(f"JSON saved: {OUTPUT_JSON}")
print(f"CSV saved:  {OUTPUT_CSV}")

print("\nData shape:")
print(df_raw.shape)

print("\nCG distribution:")
print(df_raw["ad_id"].notna().sum())

display(df_raw.head())
```

```python
import pandas as pd
import re
import numpy as np

def clean_chotot_data(df_raw):
    df_clean = pd.DataFrame()
    
    # 1. Thông tin cơ bản
    df_clean['platform'] = 'Chợ Tốt'
    df_clean['listing_id'] = df_raw.get('list_id', np.nan)
    df_clean['title'] = df_raw.get('subject', '')
    df_clean['district'] = df_raw.get('area_name', '')
    df_clean['ward'] = df_raw.get('ward_name', '')
    df_clean['address'] = df_raw.get('street_name', '')
    df_clean['price_vnd'] = df_raw.get('price', np.nan)
    df_clean['house_type'] = df_raw.get('category_name', '')
    df_clean['area_m2'] = df_raw.get('size', df_raw.get('area', np.nan))
    
    # Chuyển đổi body về chữ thường để dễ bóc tách
    bodies = df_raw.get('body', '').astype(str).str.lower()
    
    def extract_regex(text, pattern):
        match = re.search(pattern, text)
        return match.group(1).strip() if match else None

    # 2. Bóc tách tiền điện, nước, tiện ích bằng Regex
    elec_pattern = r'điện\s*[:=\-]?\s*(\d[\d\,\.]*\s*(?:k|đ|vnđ)?\s*(?:/\s*(?:số|kwh|kw|tháng|người))?)'
    df_clean['electric_price'] = bodies.apply(lambda x: extract_regex(x, elec_pattern))
    
    water_pattern = r'nước\s*[:=\-]?\s*(\d[\d\,\.]*\s*(?:k|đ|vnđ)?\s*(?:/\s*(?:khối|m3|người|ng|tháng|phòng))?)'
    df_clean['water_price'] = bodies.apply(lambda x: extract_regex(x, water_pattern))
    
    wifi_pattern = r'(?:mạng|wifi|internet)\s*[:=\-]?\s*(\d[\d\,\.]*\s*(?:k|đ|vnđ)?\s*(?:/\s*(?:tháng|phòng|người))?)'
    df_clean['wifi_price'] = bodies.apply(lambda x: extract_regex(x, wifi_pattern))
    
    vs_pattern = r'(?:vệ sinh|dịch vụ|rác)\s*[:=\-]?\s*(\d[\d\,\.]*\s*(?:k|đ|vnđ)?\s*(?:/\s*(?:tháng|người|phòng))?)'
    df_clean['other_uti'] = bodies.apply(lambda x: extract_regex(x, vs_pattern))
    
    # 3. Trích xuất tiện ích (Trạng thái True/False)
    df_clean['parking'] = bodies.str.contains(r'để xe|đỗ xe|hầm|chỗ để xe', regex=True, na=False)
    df_clean['air_condi'] = bodies.str.contains(r'điều hoà|điều hòa|máy lạnh', regex=True, na=False)
    df_clean['water_he'] = bodies.str.contains(r'nóng lạnh|bình nóng lạnh', regex=True, na=False)
    df_clean['refrigerat'] = bodies.str.contains(r'tủ lạnh', regex=True, na=False)
    df_clean['washing'] = bodies.str.contains(r'máy giặt', regex=True, na=False)
    df_clean['elevator'] = bodies.str.contains(r'thang máy', regex=True, na=False)
    df_clean['balcony'] = bodies.str.contains(r'ban công|logia', regex=True, na=False)
    df_clean['fire_safet'] = bodies.str.contains(r'pccc|phòng cháy|chữa cháy', regex=True, na=False)
    df_clean['pet_allow'] = bodies.str.contains(r'thú cưng|nuôi chó|nuôi mèo|pet', regex=True, na=False)
    
    def get_amenities_list(row):
        ams = []
        if row['air_condi']: ams.append('Điều hòa')
        if row['water_he']: ams.append('Nóng lạnh')
        if row['refrigerat']: ams.append('Tủ lạnh')
        if row['washing']: ams.append('Máy giặt')
        return ', '.join(ams)
        
    df_clean['amenities'] = df_clean.apply(get_amenities_list, axis=1)
    
    # 4. Vị trí & Liên hệ
    df_clean['latitude'] = df_raw.get('latitude', np.nan)
    df_clean['longitude'] = df_raw.get('longitude', np.nan)
    df_clean['contact_name'] = df_raw.get('full_name', df_raw.get('account_name', ''))
    
    df_clean['contact_phone'] = bodies.apply(lambda x: ', '.join(re.findall(r'\b(?:0|\+84)[1-9][0-9]{8}\b', x)) if isinstance(x, str) else '')
    df_clean['contact_zalo'] = bodies.str.contains(r'zalo', regex=True, na=False)
    
    # 5. Hình ảnh & Link
    df_clean['image_co'] = df_raw.get('number_of_images', 0)
    df_clean['image_u'] = df_raw.get('images', '')
    df_clean['listing_ur'] = 'https://nha.chotot.com/thue-phong-tro/' + df_raw.get('list_id', '').astype(str)
    
    # 6. Cột Body gốc để đối chiếu
    df_clean['body'] = df_raw.get('body', '')
    
  
    df_clean = df_clean.dropna(subset=['electric_price', 'water_price'], how='any')
    
    return df_clean

if __name__ == '__main__':
    # Đọc file raw
    df_raw = pd.read_csv('chotot_hanoi_raw.csv')
    
    # Gọi hàm clean
    df_cleaned = clean_chotot_data(df_raw)
    
    # Xuất ra file CSV
    df_cleaned.to_csv('chotot_hanoi_cleaned.csv', index=False, encoding='utf-8-sig')
    print("Đã xử lý xong, xóa các dòng thiếu thông tin điện/nước và lưu vào file chotot_hanoi_cleaned.csv!")
```
